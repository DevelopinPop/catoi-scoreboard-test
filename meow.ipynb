```python
import sqlite3
import csv
from sqlite3 import Error
```

```python
sql = ["""

CREATE TABLE `delegations` (
	`year`	INTEGER,
	`userid`	INTEGER NOT NULL,
	`role`	INTEGER,
	`school`	INTEGER,
	`id`	INTEGER PRIMARY KEY AUTOINCREMENT,
	FOREIGN KEY(`userid`) REFERENCES `people`(`userid`),
	FOREIGN KEY(`year`) REFERENCES `olympiads`(`year`)
);

""", """

CREATE TABLE `olympiads` (
	`year`	INTEGER UNIQUE,
	`date`	TEXT,
	`max_score`	INTEGER NOT NULL DEFAULT 0,
	`num_qns`	INTEGER NOT NULL DEFAULT 0,
	PRIMARY KEY(`year`)
);

""","""

CREATE TABLE `people` (
	`name`	TEXT,
	`userid`	INTEGER PRIMARY KEY AUTOINCREMENT,
	`codebreaker`	TEXT,
	`codeforces`	TEXT
);

""", """
CREATE TABLE `scores` (
	`id`	INTEGER PRIMARY KEY AUTOINCREMENT,
	`year`	INTEGER NOT NULL,
	`userid`	INTEGER NOT NULL,
	`q1`	INTEGER DEFAULT 0,
	`q2`	INTEGER DEFAULT 0,
	`q3`	INTEGER DEFAULT 0,
	`q4`	INTEGER DEFAULT 0,
	`q5`	INTEGER DEFAULT 0,
	`q6`	INTEGER DEFAULT 0,
	`q7`	INTEGER DEFAULT 0,
	`q8`	INTEGER DEFAULT 0,
	`q9`	INTEGER DEFAULT 0,
	`q10`	INTEGER DEFAULT 0,
	`award`	TEXT,
	`special_award`	TEXT,
	FOREIGN KEY(`year`) REFERENCES `olympiads`(`year`)
);

""","""


CREATE TABLE `tasks` (
	`year`	INTEGER,
	`tasknum`	INTEGER,
	`taskname`	TEXT,
	`taskid`	TEXT,
	`max_score`	INTEGER NOT NULL,
	PRIMARY KEY(`year`,`tasknum`)
);

"""]








conn = sqlite3.connect("database.db")

for i in sql:
    try:
        conn.execute(i)
    except Error as e:
        print(e)
conn.close()
```

```python
db = "database.db"

def import_people(namelist):
    # ignores first line of csv file
    conn = sqlite3.connect(db)
    with open(namelist, "r") as csvfile:
        csvreader = csv.reader(csvfile)
        next(csvreader)
        for row in csvreader:
            try: 
                conn.execute("""
INSERT INTO people(name, codebreaker) VALUES(?,?)
                
                """,(row[0],row[1]))
            except Error as e:
                print(e)
    conn.commit()
    conn.close()

    
def import_delegation(namelist):
    # ignores first line of csv file
    conn = sqlite3.connect(db)
    with open(namelist, "r") as csvfile:
        csvreader = csv.reader(csvfile)
        next(csvreader)
        for row in csvreader:
            try: 
                res = (conn.execute("""      
SELECT COUNT(people.userid) FROM people WHERE people.name == ?       
""",(row[0],))).fetchall()[0][0]
                if res == 0:
                    print("Failed to add person", row[0], ": No such person in people")
                else:
                    res = conn.execute("""      
SELECT people.userid FROM people WHERE people.name == ?    
""",(row[0],)).fetchall()[0][0]
                    conn.execute("""
INSERT INTO delegations(year, userid, role, school_code) VALUES(?,?,?,?)
                    """,(row[2],res,row[3],row[1],))
            except Error as e:
                print(e)
    conn.commit()
    conn.close()
    
    
def import_scoreboard(csv_file, year, num_qns):
    # ignores first line of csv file
    conn = sqlite3.connect(db)
    with open(csv_file, "r") as csvfile:
        csvreader = csv.reader(csvfile)
        header = next(csvreader)
        if (header[-1]=="P" and header[-2] == "Global") or (header[-3]=="P" and header[-4] == "Global") or (header[-2]=="P" and header[-3] == "Global"):
            
            for row in csvreader:
                try: 
                    res = (conn.execute("""      
    SELECT COUNT(people.userid) FROM people WHERE people.name == ?       
    """,(row[1],))).fetchall()[0][0]
                    if res == 0:
                        print("Failed to add person", row[1], ": No such person in people")
                    else:
                        res = conn.execute("""      
    SELECT people.userid FROM people WHERE people.name == ?    
    """,(row[1],)).fetchall()[0][0]
                        conn.execute("""
    INSERT INTO scores(year, userid, """ + ", ".join(["q" + str(i) for i in range()]) + """) VALUES(?,?,?,?)
                        """,(row[2],res,row[3],row[1],) )
                except Error as e:
                    print(e)
        else:
            for row in csvreader:
                try: 
                    res = (conn.execute("""      
    SELECT COUNT(people.userid) FROM people WHERE people.name == ?       
    """,(row[1],))).fetchall()[0][0]
                    if res == 0:
                        print("Failed to add person", row[1], ": No such person in people")
                    else:
                        res = conn.execute("""      
    SELECT people.userid FROM people WHERE people.name == ?    
    """,(row[1],)).fetchall()[0][0]
                        conn.execute("""
    INSERT INTO scores(year, userid, school_code, id, award, """ + ", ".join(["q" + str(i+1) for i in range(num_qns)]) + """) VALUES(?,?,?,?,?,""" + ",".join(["?" for i in range(num_qns)]) + """)
                        """,[year,res,row[-1],row[0],row[-2]] + list(row[2:2+num_qns]))
                except Error as e:
                    print(e)
    conn.commit()
    conn.close()
    
# def special_award(name, year, award):
    
#import_people("people.csv")
#import_delegation("fastimportdel.csv")
import_scoreboard("nyoi2024.csv", 2025, 8)
```
